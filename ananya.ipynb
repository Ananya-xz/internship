```python
#13
def grade(scores_values):
  if scores_values>=90:
    return("distinction")
  elif 75<=scores_values<=89:
    return("first class")
  elif 60<=scores_values<=74:
    return("second class")
  else:
    return("fail")


grades=scores.apply(grade)
print(grades)

#14
print(grades.unique())
print(scores.unique(normalize=true))
```
